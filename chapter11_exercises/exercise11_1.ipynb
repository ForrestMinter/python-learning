```python
# 作业1：生成随机整数文件
import random

print("=" * 40)
print("作业1: 生成随机整数文件")
print("=" * 40)

data = [random.randint(1, 100) for _ in range(200)]

with open('data_test.txt', 'w', encoding='utf-8') as f:
    for num in data:
        f.write(str(num) + '\n')

print("生成的数据如下：")
for i, num in enumerate(data, 1):
    print(f"{num:3d}", end=' ')
    if i % 10 == 0:
        print()
print(f"\n共生成 {len(data)} 行数据，已保存到 data_test.txt")
```

```python
# 作业2：商品数据处理
import json

print("=" * 40)
print("作业2: 商品数据处理")
print("=" * 40)

data_str = "orange 20 5 book 100 3 pen 2 20 ball 15 4"
items = data_str.split()

goods = []
for i in range(0, len(items), 3):
    goods.append({
        'name': items[i],
        'price': int(items[i+1]),
        'amount': int(items[i+2])
    })

print("(1) 构建的数据：")
for g in goods:
    print(g)

with open('goods.json', 'w', encoding='utf-8') as f:
    json.dump(goods, f, ensure_ascii=False, indent=4)
print("\n(2) 已保存到 goods.json")

with open('goods.json', 'r', encoding='utf-8') as f:
    loaded_goods = json.load(f)

total = sum(item['price'] * item['amount'] for item in loaded_goods)
print(f"\n(3) 所有商品总金额：{total} 元")
```

```python
# 作业3：等额本金还款计算
print("=" * 40)
print("作业3: 等额本金还款计算")
print("=" * 40)

def g(capital, year, rate_m):
    months = year * 12
    monthly_principal = capital / months  
    payments = []
    for i in range(months):
        interest = (capital - i * monthly_principal) * rate_m
        payment = monthly_principal + interest
        payments.append(payment)
    return payments

input_str = input("请输入贷款年数、贷款本金（万元）、年利率（%），逗号间隔：")
parts = input_str.split(',')
year = int(parts[0])
capital_wan = float(parts[1])
annual_rate = float(parts[2])

capital = capital_wan * 10000
monthly_rate = annual_rate / 100 / 12

payments = g(capital, year, monthly_rate)
total_interest = sum(payments) - capital  

with open('out.txt', 'w', encoding='utf-8') as f:
    f.write(f"等额本金共还款利息{total_interest:.2f}（元）\n")
    f.write("等额本金第一年每期还款（单位：元）如下：\n")
    for i in range(12):
        f.write(f"第{i+1}期，{payments[i]:.2f}（元）\n")

print("计算结果已保存到 out.txt")
```

```python
# 作业4：气象数据处理
import csv
import json

print("=" * 40)
print("作业4: 气象数据处理")
print("=" * 40)

input_file = "data10.csv"
output_csv = "daily_avg.csv"
output_json = "daily_avg.json"
records = []

with open(input_file, 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        year = row['year']
        month = row['month']
        mday = row['mday']
        tmax = float(row['tmax'])
        tmin = float(row['tmin'])
        tavg = round((tmax + tmin) / 2, 2)   
        records.append({
            'year': year,
            'month': month,
            'mday': mday,
            'tavg': tavg
        })
with open(output_csv, 'w', encoding='utf-8', newline='') as f:
    fieldnames = ['year', 'month', 'mday', 'tavg']
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(records)
with open(output_json, 'w', encoding='utf-8') as f:
    json.dump(records, f, ensure_ascii=False, indent=4)

print(f"已生成 {output_csv} 和 {output_json}，共处理 {len(records)} 条记录。")
```
