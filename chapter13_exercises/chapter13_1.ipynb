```python
# 作业1：动物类
class Animal:
    def __init__(self, name, age, color, food):
        self.name = name
        self.age = age
        self.color = color
        self.food = food
    
    def run(self):
        print(f"{self.name}正在奔跑")
    
    def get_age(self):
        print(f"{self.name}今年{self.age}岁了")
    
    def eat(self):
        print(f"{self.name}正在吃{self.food}")

panda = Animal("熊猫盼盼", 4, "黑白", "竹子")
tiger = Animal("老虎威威", 5, "橙黄", "肉类")
rabbit = Animal("兔子跳跳", 2, "白色", "胡萝卜")

animals = [panda, tiger, rabbit]

print("=" * 40)
print("题目1: 动物类 Animal")
print("=" * 40)

for animal in animals:
    animal.run()
    animal.get_age()
    animal.eat()
    print("-" * 30)
```

```python
# 作业2: 学生类
class Student:
    id_counter = 1
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.id = Student.id_counter
        Student.id_counter += 1
    
    def info(self):
        print(f"姓名: {self.name}, 年龄: {self.age}, 学号: {self.id}")

print("=" * 40)
print("题目2: 学生类 Student")
print("=" * 40)

s1 = Student("李晓明", 19)
s1.info()

s2 = Student("马红", 20)
s2.info()

s3 = Student("张刚", 18)
s3.info()
```

```python
# 作业3: 汽车类
class Car:
    def move(self):
        print(f"汽车型号: {self.model}, 颜色: {self.color}, 马力: {self.horsepower}匹, 正在行驶中...")

print("=" * 40)
print("题目3: 汽车类 Car")
print("=" * 40)

car_1 = Car()
car_1.color = "红色"
car_1.horsepower = 150
car_1.model = "丰田卡罗拉"

car_2 = Car()
car_2.color = "黑色"
car_2.horsepower = 200
car_2.model = "宝马X5"

car_1.move()
car_2.move()
```

```python
# 作业4: 三角形类
import math

class Triangle:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    
    def perimeter(self):
        """计算周长"""
        return self.a + self.b + self.c
    
    def area(self):
        """使用海伦公式计算面积"""
        p = self.perimeter() / 2
        s = math.sqrt(p * (p - self.a) * (p - self.b) * (p - self.c))
        return s

print("=" * 40)
print("题目4: 三角形类 Triangle")
print("=" * 40)

tri = Triangle(3, 4, 5)
print(f"三边长度: a={tri.a}, b={tri.b}, c={tri.c}")
print(f"周长: {tri.perimeter()}")
print(f"面积: {tri.area():.2f}")
```
