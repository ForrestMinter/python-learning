```python
# 作业1: 农作物类定义与生长模拟
class Crop:
    def __init__(self, name, planting_date):
        self.name = name
        self.planting_date = planting_date
        self.growth_stage = "播种"
        self.water_amount = 0
    
    def water(self, days):
        """根据浇水天数增加累计浇水量，每天5mm，并更新生长阶段"""
        self.water_amount += days * 5
        
        # 更新生长阶段
        if self.water_amount >= 50:
            self.growth_stage = "抽穗期"
        elif self.water_amount >= 20:
            self.growth_stage = "分蘖期"
    
    def get_current_status(self):
        """打印当前作物状态"""
        print(f"作物名称: {self.name}")
        print(f"生长阶段: {self.growth_stage}")
        print(f"累计浇水量: {self.water_amount}mm")

print("=" * 40)
print("补充题1: 农作物类定义与生长模拟")
print("=" * 40)

# 实例化水稻对象
rice = Crop("水稻", "2026-03-15")

# 模拟三次浇水
water_days = [3, 5, 4]
for i, days in enumerate(water_days, 1):
    print(f"--- 第{i}次浇水 ({days}天) ---")
    rice.water(days)
    rice.get_current_status()
```

```python
# 作业2: 土壤养分监测类设计
class SoilMonitor:
    def __init__(self, n, p, k):
        """初始化土壤养分含量 (mg/kg)
        n: 氮, p: 磷, k: 钾
        """
        self.n = n
        self.p = p
        self.k = k
    
    def apply_fertilizer(self, n_add, p_add, k_add):
        """施加肥料，增加对应养分含量"""
        self.n += n_add
        self.p += p_add
        self.k += k_add
        print(f"施肥完成: N+{n_add}, P+{p_add}, K+{k_add}")
    
    def check_deficiency(self):
        """判断养分是否缺乏
        氮<80mg/kg、磷<20mg/kg、钾<100mg/kg 任意满足即为缺乏
        """
        if self.n < 80 or self.p < 20 or self.k < 100:
            return True
        return False
    
    def get_soil_nutrient(self):
        """返回当前三种养分含量字典"""
        return {
            "氮(N)": self.n,
            "磷(P)": self.p,
            "钾(K)": self.k
        }

print("=" * 40)
print("补充题2: 土壤养分监测类设计")
print("=" * 40)

# 创建土壤监测对象，初始N=65、P=25、K=90
soil = SoilMonitor(65, 25, 90)
print("初始养分含量:", soil.get_soil_nutrient())
print(f"初始是否缺乏: {soil.check_deficiency()}")

# 施加尿素补充30单位氮
soil.apply_fertilizer(30, 0, 0)

print("施肥后养分含量:", soil.get_soil_nutrient())
print(f"施肥后是否仍然缺乏: {soil.check_deficiency()}")
```

```python
# 作业3: 温室环境控制类
class Greenhouse:
    def __init__(self, temperature, humidity):
        self.temperature = temperature
        self.humidity = humidity
        self.ventilation_status = False  # 通风状态，默认关闭
    
    def adjust_temp(self, target_temp):
        """根据目标温度调整环境
        如果当前温度高于目标温度5°C以上，打开通风，温度下降3°C
        如果低于目标温度，关闭通风
        """
        if self.temperature - target_temp > 5:
            self.ventilation_status = True
            self.temperature -= 3
            print(f"温度过高，已打开通风，温度下降3°C")
        elif self.temperature < target_temp:
            self.ventilation_status = False
            print(f"温度低于目标，已关闭通风")
        else:
            print(f"温度适宜，无需调整")
    
    def get_status(self):
        """返回当前环境参数描述"""
        vent = "开启" if self.ventilation_status else "关闭"
        return (
            f"当前温室状态: "
            f"  温度: {self.temperature}°C"
            f"  湿度: {self.humidity}%"
            f"  通风状态: {vent}"
        )

print("=" * 40)
print("补充题3: 温室环境控制类")
print("=" * 40)

# 实例化温室，初始温度28°C，湿度65%
gh = Greenhouse(28, 65)
print("初始状态:")
print(gh.get_status())

# 将目标温度调整为22°C
print("--- 调整温度至22°C ---")
gh.adjust_temp(22)

print("调整后的状态:")
print(gh.get_status())
```
