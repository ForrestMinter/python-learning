```python
import numpy as np
import matplotlib.pyplot as plt

print("=" * 50)
print("题目1: 方波傅里叶级数绘制")
print("=" * 50)

# (1) 生成数组t，取值范围[0, 4π)，间隔0.01
t = np.arange(0, 4 * np.pi, 0.01)

# (2) 计算k=1,2,...,10前10项的方波分项并绘制
plt.figure(figsize=(10, 6))

f_total = np.zeros_like(t)
for k in range(1, 11):
    # 方波分项: 4*sin((2k-1)t) / ((2k-1)*π)
    fk = 4 * np.sin((2 * k - 1) * t) / ((2 * k - 1) * np.pi)
    f_total += fk
    plt.plot(t, fk, label=f'k={k}', linewidth=1.5)

plt.xlabel('t')
plt.ylabel('f')
plt.title('方波傅里叶级数前10项分项')
plt.legend(loc='upper right', fontsize=8)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# (3) 将前10项相加，绘制合成方波
plt.figure(figsize=(10, 6))
plt.plot(t, f_total, 'b-', linewidth=1.5)
plt.xlabel('x')
plt.ylabel('f')
plt.title('k=10')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"\n方波级数前10项合成完成，t数组长度: {len(t)}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

print("=" * 50)
print("题目2: 气象数据可视化")
print("=" * 50)

# 读取真实气象数据
df = pd.read_csv('data10.csv')
print(f"数据行数: {len(df)}")
print(df.head())

# ============================================================
# (1) srad、RH、wind 用折线图分别绘制（3个子图）
# ============================================================
fig, axes = plt.subplots(3, 1, figsize=(10, 12))

# srad - 向下短波辐射
axes[0].plot(df.index, df['srad'], linewidth=1.2)
axes[0].set_title('Downward shortwave radiation', fontsize=14, fontweight='bold')
axes[0].set_ylabel('Downward shortwave radiation (W/m2)')
axes[0].set_xlabel('')

# RH - 相对湿度
axes[1].plot(df.index, df['RH'], linewidth=1.2)
axes[1].set_title('Relative humidity', fontsize=14, fontweight='bold')
axes[1].set_ylabel('Relative humidity (%)')
axes[1].set_xlabel('')

# wind - 风速
axes[2].plot(df.index, df['wind'], linewidth=1.2)
axes[2].set_title('Wind speed', fontsize=14, fontweight='bold')
axes[2].set_ylabel('Wind speed (m/s)')
axes[2].set_xlabel('')

plt.tight_layout()
plt.show()

print("\n(1) srad、RH、wind 折线图绘制完成")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('data10.csv')

# ============================================================
# (2) prec 用柱状图绘制
# ============================================================
plt.figure(figsize=(10, 6))
plt.bar(df.index, df['prec'], width=1.0, color='steelblue', edgecolor='none')
plt.title('Precipitation rate', fontsize=14, fontweight='bold')
plt.xlabel('Day of year')
plt.ylabel('Precipitation rate (mm/day)')
plt.xlim(-10, len(df) + 10)
plt.tight_layout()
plt.show()

print("(2) prec 柱状图绘制完成")

# ============================================================
# (3) tmax和tmin在同一张图上用折线图绘制，中间区域用50%透明度灰色填充
# ============================================================
plt.figure(figsize=(10, 6))
plt.plot(df.index, df['tmax'], label='Daily maximum temperature', linewidth=1.5)
plt.plot(df.index, df['tmin'], label='Daily minimum temperature', linewidth=1.5)

# 填充两条曲线之间的区域
plt.fill_between(df.index, df['tmin'], df['tmax'], color='grey', alpha=0.5)

plt.title('Air temperature', fontsize=14, fontweight='bold')
plt.xlabel('Day of year')
plt.ylabel('Air temperature (C)')
plt.legend(loc='lower center')
plt.xlim(-10, len(df) + 10)
plt.tight_layout()
plt.show()

print("(3) tmax和tmin折线图+灰色填充绘制完成")

# ============================================================
# (4) tmin和tmax关系散点图，用红色虚线画出1:1线
# ============================================================
plt.figure(figsize=(8, 8))
plt.scatter(df['tmax'], df['tmin'], s=8, c='black', marker='.', label='Tmax v.s. Tmin')

# 1:1线
min_val = min(df['tmin'].min(), df['tmax'].min()) - 5
max_val = max(df['tmin'].max(), df['tmax'].max()) + 5
plt.plot([min_val, max_val], [min_val, max_val], 'r--', linewidth=1.5, label='1:1 line')

plt.title('Air temperature', fontsize=14, fontweight='bold')
plt.xlabel('Daily maximum temperature (C)')
plt.ylabel('Daily minimum temperature (C)')
plt.legend(loc='upper left')
plt.axis('equal')
plt.tight_layout()
plt.show()

print("(4) tmin-tmax散点图+1:1线绘制完成")
```
