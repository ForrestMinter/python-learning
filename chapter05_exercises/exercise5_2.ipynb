```python
# 作业1:小麦病害数据列表操作
print("=" * 40)
print("作业1:小麦病害数据列表操作")
print("=" * 40)

wheat_diseases = [15.2, 8.7, 22.1, 8.7, 15.2, 10.5] 
print("小麦病害数据:", wheat_diseases)

## （1）
wheat_diseases.pop()
print("删除后的数据：", wheat_diseases)

## （2）
wheat_diseases.insert(0, 5.3)
print("插入后的数据：", wheat_diseases)

## （3）
wheat_diseases.sort()
print("排序后的数据：", wheat_diseases)
```

```python
# 作业2:土壤样本元组操作
print("\n" + "=" * 40)
print("作业2:土壤样本元组操作")
print("=" * 40)

sample1 = ('S001', 6.8, 2.4)
sample2 = ('S002', 7.5, 1.8)
sample3 = ('S003', 5.9, 3.1)
print("sample1:", sample1)
print("sample2:", sample2)
print("sample3:", sample3)

## (1)
all_samples = (sample1, sample2, sample3)
print("(1) all_samples:", all_samples)

## (2)
print("(2) 尝试修改sample2的pH值:")
try:
    sample2[1] = 7.2
except TypeError as e:
    print(f"报错: {e}")
    print("原因：元组是不可变类型，创建后不能修改元素")

## (3)
organic_matter = all_samples[1][2]
print("(3) 第二个样本的有机质含量:", organic_matter)
```
