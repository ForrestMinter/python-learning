```python
# 作业1：矩形类的定义与实例化
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height

print("=" * 40)
print("题目1: 矩形类的定义与实例化")
print("=" * 40)

rect1 = Rectangle(5, 10)
rect2 = Rectangle(3, 7)

print(f"矩形1: 宽={rect1.width}, 高={rect1.height}, 面积={rect1.area()}")
print(f"矩形2: 宽={rect2.width}, 高={rect2.height}, 面积={rect2.area()}")
```

```python
# 作业2：继承Rectangle类实现Square类
class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)

print("=" * 40)
print("题目2: 继承实现Square类")
print("=" * 40)

sq = Square(6)
print(f"正方形: 边长={sq.width}, 面积={sq.area()}")
```

```python
# 作业3：Point类运算符重载
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f"Point({self.x},{self.y})"
    
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)
    
    def __sub__(self, other):
        return Point(self.x - other.x, self.y - other.y)
    
    def __mul__(self, n):
        return Point(self.x * n, self.y * n)
    
    def __truediv__(self, n):
        return Point(self.x / n, self.y / n)

print("=" * 40)
print("题目3: Point类运算符重载")
print("=" * 40)

a, b = Point(4, 5), Point(2, 3)
print(f"a = {a}")
print(f"b = {b}")
print(f"a + b = {a + b}")      # Point(6,8)
print(f"a - b = {a - b}")      # Point(2,2)
print(f"a * 2 = {a * 2}")      # Point(8,10)
print(f"a / 2 = {a / 2}")      # Point(2.0,2.5)
```

```python
# 作业4：用继承方法定义新类完成上一题
class PointBase:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f"Point({self.x},{self.y})"

class PointExt(PointBase):
    def __add__(self, other):
        return PointExt(self.x + other.x, self.y + other.y)
    
    def __sub__(self, other):
        return PointExt(self.x - other.x, self.y - other.y)
    
    def __mul__(self, n):
        return PointExt(self.x * n, self.y * n)
    
    def __truediv__(self, n):
        return PointExt(self.x / n, self.y / n)

print("=" * 40)
print("题目4: 继承方式实现Point运算符重载")
print("=" * 40)

a, b = PointExt(4, 5), PointExt(2, 3)
print(f"a = {a}")
print(f"b = {b}")
print(f"a + b = {a + b}")
print(f"a - b = {a - b}")
print(f"a * 2 = {a * 2}")
print(f"a / 2 = {a / 2}")
```

```python
# 作业5：图形绘制程序（Shape/Line/Circle）
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f"({self.x},{self.y})"

class Shape:
    def __init__(self):
        self.position = Point(0, 0)
    
    def draw(self):
        print(str(self))
    
    def move_to(self, position):
        print(f"moving to {position}")
        self.position = position

class Line(Shape):
    def __init__(self, p1, p2):
        super().__init__()
        self.position = p1
        self.p2 = p2
    
    def move_to(self, position):
        dx = position.x - self.position.x
        dy = position.y - self.position.y
        self.p2 = Point(self.p2.x + dx, self.p2.y + dy)
        super().move_to(position)
    
    def __str__(self):
        return f"Line:{self.position}----{self.p2}"

class Circle(Shape):
    def __init__(self, center, radius):
        super().__init__()
        self.position = center
        self.radius = radius
    
    def __str__(self):
        return f"Circle center:{self.position},R={self.radius}"

print("=" * 40)
print("题目5: 图形绘制程序")
print("=" * 40)

l = Line(Point(0, 0), Point(20, 20))
l.draw()
l.move_to(Point(3, 5))
l.draw()

c = Circle(Point(10, 10), 5)
c.draw()
c.move_to(Point(30, 20))
c.draw()
```

```python
# 作业6：农业病虫害记录类
class PestRecord:
    def __init__(self, crop_name, pest_name):
        self.crop_name = crop_name
        self.pest_name = pest_name
        self.records = []
    
    def add_occurrence(self, date_str):
        self.records.append(date_str)
    
    def count_occurrence(self):
        return len(self.records)
    
    def __str__(self):
        return f"{self.crop_name}-{self.pest_name}:共发生{self.count_occurrence()}次"

print("=" * 40)
print("题目6: 农业病虫害记录类")
print("=" * 40)

pest = PestRecord("水稻", "稻飞虱")
pest.add_occurrence("2026-05-10")
pest.add_occurrence("2026-05-25")
pest.add_occurrence("2026-06-08")

print(pest)
```
